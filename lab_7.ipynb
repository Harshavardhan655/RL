```python

# Calculate Undiscounted and Discounted Returns

# Accept number of time steps
n = int(input("Enter number of time steps: "))

# Accept rewards
rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

# Calculate undiscounted return
undiscounted_return = sum(rewards)

# Accept discount factor
gamma = float(input("Enter discount factor (gamma): "))

# Calculate discounted return
discounted_return = 0

for t in range(n):
    discounted_return += (gamma ** t) * rewards[t]

# Display results
print("\n----- RESULTS -----")
print("Entered Rewards:", rewards)
print("Undiscounted Return:", undiscounted_return)
print("Discount Factor:", gamma)
print("Discounted Return:", round(discounted_return, 2))
```

```python

%%writefile grid_env_det.py

# 2 x 3 Grid World MDP

class GridWorld:

    def __init__(self):

        # Define State Space
        self.states = {
            "S0": (0, 0),
            "S1": (0, 1),
            "G":  (0, 2),
            "S3": (1, 1),
            "S4": (1, 2)
        }

        # Define Action Space
        self.actions = ["Up", "Down", "Left", "Right"]

        # Define Rewards
        self.normal_reward = -1
        self.goal_reward = 10
        self.obstacle_reward = -5

        # Define Discount Factor
        self.gamma = 0.9

        # Define Grid
        self.grid = [
            ["S0", "S1", "G"],
            ["X",  "S3", "S4"]
        ]

        # Define Initial State
        self.current_state = "S0"

        # Terminal State
        self.terminal_states = ["G"]

    # Define State Space
    def get_states(self):
        return list(self.states.keys())

    # Define Action Space
    def get_actions(self):
        return self.actions

    # Define Reward Function
    def reward_function(self, next_state):

        if next_state == "G":
            return self.goal_reward

        return self.normal_reward

    # Define Discount Factor
    def get_discount_factor(self):
        return self.gamma

    # Define Transition Function
    def transition_function(self, state, action):

        row, col = self.states[state]

        if action == "Up":
            new_row, new_col = row - 1, col

        elif action == "Down":
            new_row, new_col = row + 1, col

        elif action == "Left":
            new_row, new_col = row, col - 1

        elif action == "Right":
            new_row, new_col = row, col + 1

        else:
            return state, self.normal_reward

        # Check grid boundaries
        if new_row < 0 or new_row >= 2 or new_col < 0 or new_col >= 3:
            return state, self.normal_reward

        # Check obstacle
        if self.grid[new_row][new_col] == "X":
            return state, self.obstacle_reward

        # Find next state
        next_state = self.grid[new_row][new_col]

        # Calculate reward
        reward = self.reward_function(next_state)

        return next_state, reward
```

```python

# Import the Grid World Environment

from grid_env_det import GridWorld

# Create Environment
env = GridWorld()

# Display State Space
print("State Space:", env.get_states())

# Display Action Space
print("Action Space:", env.get_actions())

# Display Discount Factor
print("Discount Factor:", env.get_discount_factor())

# Test Transition Function
state = "S0"

print("\nCurrent State:", state)

for action in env.get_actions():

    next_state, reward = env.transition_function(state, action)

    print("Action:", action,
          "| Next State:", next_state,
          "| Reward:", reward)
```
