```python

import gymnasium as gym
import numpy as np

def random_agent(env_name, episodes=10):
    env = gym.make(env_name)

    total_rewards = []

    print("\nEnvironment:", env_name)
    print("Action space:", env.action_space)

    for episode in range(episodes):
        state, info = env.reset()
        done = False
        total_reward = 0
        steps = 0

        while not done:
            # Select a random action
            action = env.action_space.sample()

            # Execute action in the environment
            next_state, reward, terminated, truncated, info = env.step(action)

            total_reward += reward
            steps += 1

            # Update the current state
            state = next_state

            # Stop when the episode ends
            done = terminated or truncated

        total_rewards.append(total_reward)

        print(
            f"Episode {episode + 1}: "
            f"Steps = {steps}, Reward = {total_reward}"
        )

    print("\nTotal reward:", sum(total_rewards))
    print("Average reward:", np.mean(total_rewards))

    env.close()


# Run the random agent on all three environments
random_agent("CartPole-v1", episodes=10)
random_agent("MountainCar-v0", episodes=10)
random_agent("FrozenLake-v1", episodes=10)
```
